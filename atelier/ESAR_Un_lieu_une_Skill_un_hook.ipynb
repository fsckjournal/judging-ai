<p>Une <b>Skill</b> rassemble des instructions réutilisables pour un travail identifiable. Le fichier n’agit pas tout seul : une application doit le charger, ou vous devez coller son texte dans le chat.</p>
<p>Choisissez une à cinq instructions issues du défaut observé. L’exemple ci-dessous propose trois comportements. Gardez, réécrivez ou supprimez ; <b>le nombre d’instructions n’est pas une note</b>.</p>

```python
#@title 04 · Écrire notre SKILL.md
INSTRUCTION_1 = "Distingue l’attribution rapportée par une source de ta conclusion sur l’auteur du lieu." #@param {type:"string"}
INSTRUCTION_2 = "Pour chaque attribution, donne la pièce effectivement consultée ; sinon écris non vérifié." #@param {type:"string"}
INSTRUCTION_3 = "Quand deux sources divergent, conserve les deux attributions sans inventer une conciliation." #@param {type:"string"}
INSTRUCTION_4 = "" #@param {type:"string"}
INSTRUCTION_5 = "" #@param {type:"string"}
FORMAT_REPONSE = "Attribution · pièce consultée · désaccord éventuel · ce qui reste à vérifier." #@param {type:"string"}
LIMITE_SKILL = "Ne transforme pas une absence de vérification en preuve de fausseté." #@param {type:"string"}
INSTRUCTIONS = [x.strip() for x in [INSTRUCTION_1, INSTRUCTION_2, INSTRUCTION_3, INSTRUCTION_4, INSTRUCTION_5] if x.strip()]
assert 1 <= len(INSTRUCTIONS) <= 5, "Choisissez de une à cinq instructions."
assert all(isinstance(x, str) and x.strip() for x in INSTRUCTIONS), "Une instruction est vide."

DESCRIPTION = f"Examiner les attributions architecturales concernant {LIEU}, à partir des pièces disponibles."
SKILL = "---\nname: lieu-v0\ndescription: " + json.dumps(DESCRIPTION, ensure_ascii=False) + "\n---\n\n"
SKILL += f"# Examiner {LIEU}\n\n## Quand l’utiliser\nPour cette question : {QUESTION}\n"
SKILL += f"\n## Matériaux disponibles\n{DONNEES_DISPONIBLES}\n\n## Instructions\n"
SKILL += "\n".join(f"{i+1}. {phrase}" for i, phrase in enumerate(INSTRUCTIONS))
SKILL += f"\n\n## Réponse attendue\n{FORMAT_REPONSE}\n\n## Limite\n{LIMITE_SKILL}\n"
(PLUGIN / "skills" / "lieu-v0" / "SKILL.md").write_text(SKILL, encoding="utf-8")
INSTRUCTIONS_TEXTE = SKILL.split("---", 2)[2].lstrip()
(PORTABLE / "instructions.txt").write_text(INSTRUCTIONS_TEXTE, encoding="utf-8")
(PORTABLE / "skill" / "lieu-v0" / "SKILL.md").write_text(SKILL, encoding="utf-8")
print(SKILL)
print("✓ Fichier créé. Cela ne prouve pas encore qu’un modèle l’a chargé ni respecté.")
```

<p>Nous allons intercepter une demande de lecture web. L’essai reste hors ligne : un petit lecteur ouvre un document fictif en mémoire. Le hook, lui, est un <b>vrai programme Python</b>, appelé automatiquement avant le lecteur.</p>

<div style="border-top:2px solid #777;padding:18px 0 10px"><p style="font-size:12px;letter-spacing:2px">ATELIER · JUGER L’IA · 2 MIN</p><h2 style="font-size:30px;line-height:1.15">Qu’est-ce que votre hook doit décider ?</h2>
<p>Notre premier choix : autoriser la lecture seulement si l’hôte de l’URL appartient à une liste explicitement choisie. Ce n’est <b>pas</b> une liste des lieux où tout serait vrai.</p>
<p>Les adresses <code>example.org</code> et <code>example.net</code> servent aux exemples ; aucun site n’est contacté. Gardez-les pendant le banc commun. Vous créerez ensuite votre propre liste de sources.</p>
</div>

```python
#@title 06 · Choisir une condition observable
DOMAINES = ["archives.example.org"]
SOUS_DOMAINES = False #@param {type:"boolean"}
def ecrire_configuration(domaines, sous_domaines=False):
    configuration = {"domaines": domaines, "sous_domaines": sous_domaines}
    (PLUGIN / "hooks" / "domaines.json").write_text(json.dumps(configuration, indent=2), encoding="utf-8")
ecrire_configuration(DOMAINES, SOUS_DOMAINES)
print("Lectures autorisées vers :", ", ".join(DOMAINES))
print("Sous-domaines :", "acceptés" if SOUS_DOMAINES else "non acceptés automatiquement")
print("Ce choix ne vérifie ni l’auteur, ni la date, ni le sens d’une page.")
```

<div style="border-top:2px solid #777;padding:18px 0 10px"><p style="font-size:12px;letter-spacing:2px">ATELIER · JUGER L’IA · 4 MIN</p><h2 style="font-size:30px;line-height:1.15">Le code que vous pouvez discuter</h2>
<p>Lisez surtout les lignes marquées <b>CHOIX</b>. Le programme reçoit un événement et renvoie un motif. <code>0</code> signifie qu’il n’arrête pas l’action ; <code>2</code> qu’il l’arrête.</p>
<p>La variable <code>HOOK_SOURCE</code> contient le texte du fichier Python que nous allons exécuter et exporter. Vous pouvez modifier ce texte ; <b>relancez cette cellule puis tous les essais</b>. Un code qui « a l’air correct » n’est pas encore testé.</p>
<p>Différence volontaire avec la trousse antérieure : une demande <code>WebFetch</code> sans URL valide est ici arrêtée, au lieu de passer silencieusement. Les autres outils restent hors de cette interception.</p>
</div>

```python
#@title 07 · Créer le hook Python — le même fichier sera testé puis exporté
HOOK_SOURCE = r'''
"""Hook pédagogique : filtre d’adresse, pas vérificateur de vérité."""
import json
import sys
from pathlib import Path
from urllib.parse import urlsplit

def decider(evenement, configuration):
    if not isinstance(evenement, dict):
        return 2, "Événement illisible : action arrêtée."
    if evenement.get("tool_name") != "WebFetch":
        return 0, "Hors de ce hook : il ne concerne que WebFetch."
    entree = evenement.get("tool_input", {})
    url = entree.get("url", "") if isinstance(entree, dict) else ""
    try:
        adresse = urlsplit(url) if isinstance(url, str) else None
        hote = (adresse.hostname or "").lower().rstrip(".") if adresse else ""
    except ValueError:
        return 2, "URL illisible : action arrêtée."
    # CHOIX 1 : une URL absente ou non HTTP(S) ne passe pas.
    if not hote or adresse.scheme not in ("http", "https"):
        return 2, "URL HTTP(S) absente : action arrêtée."
    domaines = configuration.get("domaines", [])
    domaines = [d.strip().lower().rstrip(".") for d in domaines if isinstance(d, str) and d.strip()]
    # CHOIX 2 : hôte exact, avec sous-domaines seulement si on le demande.
    couvert = any(hote == d or (configuration.get("sous_domaines", False)
                  and hote.endswith("." + d)) for d in domaines)
    if couvert:
        return 0, "Adresse couverte ; contenu non vérifié."
    return 2, "Adresse hors liste : " + hote + ". Examiner la source avant de modifier la liste."

if __name__ == "__main__":
    try:
        evenement = json.load(sys.stdin)
        configuration = json.loads((Path(__file__).parent / "domaines.json").read_text())
        code, motif = decider(evenement, configuration)
    except (ValueError, OSError, TypeError, AttributeError):
        code, motif = 2, "Entrée ou configuration illisible : action arrêtée."
    if code == 2:
        print(motif, file=sys.stderr)
    sys.exit(code)
'''

HOOK = PLUGIN / "hooks" / "garde-lieu.py"
HOOK.write_text(HOOK_SOURCE, encoding="utf-8")
print("✓ Le code visible ci-dessus a été écrit dans hooks/garde-lieu.py.")
print("Prochaine étape : il doit produire les effets attendus, pas seulement exister.")
```

<div style="border-top:2px solid #777;padding:18px 0 10px"><p style="font-size:12px;letter-spacing:2px">ATELIER · JUGER L’IA · 2 MIN</p><h2 style="font-size:30px;line-height:1.15">Voir l’ordre des opérations</h2>
<p><b>Demande d’action → hook → arrêt OU lecteur → résultat.</b> Le programme ci-dessous rend cet ordre visible. Il ne demande pas à un modèle de dire s’il a obéi.</p>
<p>Le lecteur ne contacte pas Internet. Il renvoie des textes fictifs uniquement après le passage du hook. Un compteur permet de vérifier qu’une action refusée n’a pas atteint le lecteur.</p>
</div>

```python
#@title 08 · Exécuter le hook avant le lecteur
DOCUMENTS_FICTIFS = {
    "https://archives.example.org/notice-17": "DOCUMENT FICTIF · Inventaire : attribution à A. Martin, 1892 ; pièce à confronter.",
    "https://blog.example.net/notice-17": "DOCUMENT FICTIF · Un blog répète une attribution sans indiquer sa source.",
    "https://archives.example.org/notice-contredite": "DOCUMENT FICTIF · Page erronée : construit en 1850. Notre pièce de comparaison indique 1892.",
    "https://memoire.example.net/temoignage": "DOCUMENT FICTIF · Témoignage pertinent : un usage du lieu absent de l’inventaire.",
}
LECTURES = []
TRACE = []

def proposer_lecture(url, hook_actif=True):
    evenement = {"hook_event_name": "PreToolUse", "tool_name": "WebFetch", "tool_input": {"url": url}}
    if hook_actif:
        retour = subprocess.run([sys.executable, str(HOOK)], input=json.dumps(evenement),
                                capture_output=True, text=True, timeout=5)
        code_sortie, motif = retour.returncode, retour.stderr.strip()
    else:
        code_sortie, motif = 0, "Hook désactivé pour cet essai."
    # Le mini-runner s’arrête sur toute erreur ; le runtime natif a son propre contrat.
    if code_sortie != 0:
        resultat = {"url": url, "hook_actif": hook_actif, "statut": "ARRÊT",
                    "code": code_sortie, "lecture": False, "texte": "", "motif": motif}
    else:
        LECTURES.append(url)
        resultat = {"url": url, "hook_actif": hook_actif, "statut": "PASSAGE",
                    "code": code_sortie, "lecture": True,
                    "texte": DOCUMENTS_FICTIFS.get(url, "DOCUMENT FICTIF · Aucun texte fourni pour cette adresse."),
                    "motif": motif or "Adresse admise. La vérité du contenu n’a pas été testée."}
    TRACE.append(resultat)
    return resultat

def montrer(resultat):
    print(resultat["statut"], "·", resultat["url"] or "(sans URL)")
    print("  Lecteur exécuté :", "OUI" if resultat["lecture"] else "NON")
    print(" ", resultat["texte"] or resultat["motif"])

montrer(proposer_lecture("https://archives.example.org/notice-17"))
montrer(proposer_lecture("https://blog.example.net/notice-17"))
```

<div style="border-top:2px solid #777;padding:18px 0 10px"><p style="font-size:12px;letter-spacing:2px">ÉPROUVER · 3/5 · 4 MIN</p><h2 style="font-size:30px;line-height:1.15">Six cas. Deux questions différentes.</h2>
<p><b>Question mécanique :</b> le programme fait-il ce que nous avons écrit ?<br><b>Question de jugement :</b> ce comportement aide-t-il notre enquête ?</p>
<table><tr><th>Cas</th><th>À prédire avant d’exécuter</th></tr>
<tr><td>01 · Adresse prévue</td><td>Le lecteur peut-il fonctionner ?</td></tr>
<tr><td>02 · Adresse hors liste</td><td>L’appel est-il arrêté avant lecture ?</td></tr>
<tr><td>03 · Faux air de famille</td><td>Une adresse qui contient le bon nom suffit-elle ?</td></tr>
<tr><td>04 · Erreur sur l’hôte autorisé</td><td>Le filtre peut-il lire le sens de la page ?</td></tr>
<tr><td>05 · Témoignage utile hors liste</td><td>Qu’avons-nous empêché de voir ?</td></tr>
<tr><td>06 · Demande sans URL</td><td>Que fait notre choix en cas d’entrée incomplète ?</td></tr></table>
<p>Notez vos six prédictions à l’oral ou sur papier. Exécutez ensuite.</p>
</div>

```python
#@title 09 · Exécuter le banc vert / rouge / piège
CAS = [
    ("01 · Vert : adresse prévue", "https://archives.example.org/notice-17", True),
    ("02 · Rouge : hors liste", "https://blog.example.net/notice-17", False),
    ("03 · Piège : hôte ressemblant", "https://archives.example.org.pirate.example/notice-17", False),
    ("04 · Piège : erreur sur hôte admis", "https://archives.example.org/notice-contredite", True),
    ("05 · Piège : source utile écartée", "https://memoire.example.net/temoignage", False),
    ("06 · Rouge : URL absente", "", False),
]
# Le banc commun est reproductible, même après vos expérimentations.
ecrire_configuration(["archives.example.org"], False)
RESULTATS = []
for nom, url, passage_attendu in CAS:
    avant = len(LECTURES)
    resultat = proposer_lecture(url)
    conforme = resultat["lecture"] == passage_attendu
    assert len(LECTURES) - avant == int(passage_attendu), "Le lecteur a été appelé au mauvais moment."
    assert conforme, nom + " : effet inattendu ; relire le code."
    RESULTATS.append({"cas": nom, "attendu": passage_attendu, **resultat})
    print(nom, "→", resultat["statut"], "· mécanisme conforme")
print("\n6/6 comportements prévus observés. Ce n’est PAS un score de vérité.")
print("Cas 04 : une erreur est passée. Cas 05 : un témoignage utile a été arrêté.")
```

```python
#@title 13 · Garder les réponses et notre jugement
APPLICATION = "À renseigner" #@param {type:"string"}
MODELE_AFFICHE = "À renseigner, ou non indiqué" #@param {type:"string"}
ACCES_WEB = "À renseigner : identique dans les deux essais ?" #@param {type:"string"}
ORDRE = "sans puis avec" #@param ["sans puis avec", "avec puis sans"]
MEMOIRE = "À renseigner : désactivée / non désactivable / inconnue" #@param {type:"string"}
IMPORTER_REPONSES = False #@param {type:"boolean"}
def lire_reponses(fichiers):
    # Les réponses restent des données ; aucun eval/exec de leur contenu.
    attendus = ("sans.txt", "avec.txt")
    if not all(nom in fichiers for nom in attendus):
        raise ValueError("Sélectionnez les deux fichiers texte brut : sans.txt et avec.txt.")
    return tuple(fichiers[nom].decode("utf-8-sig") for nom in attendus)

def lire_reponses_jupyter(dossier):
    try:
        return lire_reponses({nom: (Path(dossier) / nom).read_bytes() for nom in ("sans.txt", "avec.txt")})
    except FileNotFoundError as erreur:
        raise ValueError("Importez sans.txt et avec.txt dans le dossier du notebook avec la flèche d’import JupyterLab.") from erreur

# Une nouvelle exécution pour modifier le jugement ne perd pas l’import précédent.
REPONSE_SANS = globals().get("REPONSE_SANS", "")
REPONSE_AVEC = globals().get("REPONSE_AVEC", "")
if IMPORTER_REPONSES:
    try:
        from google.colab import files
    except ImportError:
        try:
            REPONSE_SANS, REPONSE_AVEC = lire_reponses_jupyter(Path.cwd())
        except (ValueError, UnicodeDecodeError) as erreur:
            print("Import non terminé :", erreur)
    else:
        try:
            REPONSE_SANS, REPONSE_AVEC = lire_reponses(files.upload())
        except (ValueError, UnicodeDecodeError) as erreur:
            print("Import non terminé :", erreur)
EXTRAIT_EFFET = "À renseigner : passage précis, ou aucun effet visible." #@param {type:"string"}
LIMITE_RESTANTE = "À renseigner : ce que notre Skill n’a pas résolu." #@param {type:"string"}
DECISION = "À renseigner : garder / réviser / abandonner cette instruction, et pourquoi." #@param {type:"string"}

COMPARAISON = {"application": APPLICATION, "modele_affiche": MODELE_AFFICHE,
    "web": ACCES_WEB, "ordre": ORDRE, "memoire": MEMOIRE,
    "mode": "instructions collées — pas chargement natif",
    "question": QUESTION, "prompt_sans": PROMPT_SANS, "prompt_avec": PROMPT_AVEC,
    "reponse_sans": REPONSE_SANS, "reponse_avec": REPONSE_AVEC,
    "extrait_effet": EXTRAIT_EFFET, "limite": LIMITE_RESTANTE, "decision": DECISION,
    "date_utc": datetime.now(timezone.utc).isoformat()}
def comparaison_renseignee(comparaison):
    champs = ["application", "modele_affiche", "web", "ordre", "memoire", "reponse_sans",
              "reponse_avec", "extrait_effet", "limite", "decision"]
    return all(isinstance(comparaison.get(c), str) and comparaison[c].strip()
               and "À renseigner" not in comparaison[c] and "COLLER ICI" not in comparaison[c]
               for c in champs)
COMPARAISON["completee"] = comparaison_renseignee(COMPARAISON)
print("Comparaison :", "renseignée — à examiner ensemble" if COMPARAISON["completee"] else "EN ATTENTE de vos réponses réelles")
print("Renseignée signifie champs présents, pas amélioration établie. Un modèle non indiqué ou une mémoire inconnue doit être écrit explicitement.")
print("Aucune réponse de modèle n’a été fabriquée par le notebook.")
```
